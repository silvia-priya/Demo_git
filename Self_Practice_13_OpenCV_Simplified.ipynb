In this notebook we will read an image, resize it, draw shapes on it, convert it to grayscale, apply thresholding, and detect edges.

## Import necessary libraries

```python
import cv2
import matplotlib.pyplot as plt
```

## 1. Read and Display an Image

OpenCV reads colour images in **BGR** order, but Matplotlib expects **RGB**, so we convert before displaying.

```python
# Read an image using OpenCV (replace with your own image path)
img = cv2.imread('Couple.jpg')

# Convert the image from BGR to RGB
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Display the image
plt.imshow(img_rgb)
plt.title("Original Image")
plt.show()
```

## 2. Resize an Image

The size is given as **(width, height)**. Resizing every image to a fixed size is a common preprocessing step before feeding images to a CNN.

```python
img_resized = cv2.resize(img_rgb, (200, 200))

plt.imshow(img_resized)
plt.title("Resized Image")
plt.show()
```

## 3. Image Annotations

Drawing functions change the image in place, so we work on a copy. Our image is in **RGB** order, so `(255, 0, 0)` is red, `(0, 255, 0)` is green, and `(0, 0, 255)` is blue.

Points are given as **(x, y)** pixel coordinates, measured from the top-left corner. Notice the axis ticks in the plot, which help you see the coordinates.

```python
img_annotated = img_rgb.copy()

# Draw a rectangle: top-left corner, bottom-right corner, colour (red), thickness
cv2.rectangle(img_annotated, (50, 50), (150, 150), (255, 0, 0), 3)

# Draw a circle: centre, radius, colour (green), thickness
cv2.circle(img_annotated, (100, 100), 30, (0, 255, 0), 3)

plt.imshow(img_annotated)
plt.title("Annotated Image")
plt.show()
```

## 4. Convert to Grayscale and Apply Thresholding

Thresholding works on a single-channel image, so we convert to grayscale first. Every pixel brighter than 127 becomes white (255), and every other pixel becomes black (0).

```python
img_gray = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2GRAY)

_, img_thresh = cv2.threshold(img_gray, 127, 255, cv2.THRESH_BINARY)

plt.imshow(img_thresh, cmap='gray')
plt.title("Thresholded Image")
plt.show()
```

## 5. Edge Detection

Canny edge detection finds outlines where brightness changes sharply.
- Gradient above the upper threshold (200) is a definite edge.
- Gradient below the lower threshold (100) is not an edge.
- In between, a pixel is kept only if it is connected to a strong edge.

```python
edges = cv2.Canny(img_gray, 100, 200)

plt.imshow(edges, cmap='gray')
plt.title("Edge Detection")
plt.show()
```
